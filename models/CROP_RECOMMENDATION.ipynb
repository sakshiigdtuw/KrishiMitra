```python
x = df.drop('label',axis=1)
y = df['label']
```

```python
y.info()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

### Feature Scaling

It's a good practice to scale numerical features, especially for algorithms sensitive to feature magnitudes (like Logistic Regression). We'll use `StandardScaler` to transform the features such that they have a mean of 0 and a standard deviation of 1. This step will be applied after splitting the data to prevent data leakage from the test set.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

# Convert scaled arrays back to DataFrames for better readability and to retain column names
x_train_scaled = pd.DataFrame(x_train_scaled, columns=x_train.columns, index=x_train.index)
x_test_scaled = pd.DataFrame(x_test_scaled, columns=x_test.columns, index=x_test.index)

print("Scaled Training Data (first 5 rows):")
display(x_train_scaled.head())

print("\nScaled Test Data (first 5 rows):")
display(x_test_scaled.head())
```

```python
x_train.info()
```

```python
y_train.info()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000) # Increased max_iter to potentially help with convergence
model.fit(x_train_scaled, y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
lr_accuracy = accuracy_score(y_test,y_pred)
print("Logistic regression model accuracy: " + str(lr_accuracy))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

model2 = DecisionTreeClassifier()
model3 = RandomForestClassifier()
model2.fit(x_train,y_train)
model3.fit(x_train,y_train)
y_pred2 = model2.predict(x_test)
y_pred3 = model3.predict(x_test)
```

```python
dt_acc = accuracy_score(y_test,y_pred2)
rf_acc = accuracy_score(y_test,y_pred3)
print("Decision tree model accuracy: " + str(dt_acc))
print("Random forest model accuracy: " + str(rf_acc))
```

Evaluate the trained models (Logistic Regression, Decision Tree, Random Forest) using additional classification metrics such as Precision, Recall, F1-score, and display a classification report and confusion matrix for each model. ## Hyperparameter Tuning for Random Forest

Optimize the hyperparameters of the Random Forest Classifier, which showed the highest accuracy, using techniques like GridSearchCV or RandomizedSearchCV to find the best model configuration.


**Reasoning**:
I need to set up the hyperparameter tuning for the Random Forest model. This first step will involve importing the necessary libraries, defining a parameter grid for the `RandomForestClassifier`, and initializing `GridSearchCV`.



```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Define the parameter grid for Random Forest
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_features': [None, 'sqrt', 'log2'], # Changed 'auto' to None
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10]
}

# Initialize GridSearchCV with the RandomForestClassifier
grid_search = GridSearchCV(estimator=model3, param_grid=param_grid, cv=5, n_jobs=-1, verbose=2)

print("GridSearchCV initialized with parameter grid:")
for param, values in param_grid.items():
    print(f"  {param}: {values}")

print("Starting GridSearchCV fit...")
grid_search.fit(x_train_scaled, y_train)
print("GridSearchCV fit complete.")
```

**Reasoning**:
Fit the `GridSearchCV` object to the training data. This will perform the exhaustive search over the defined parameter grid.



```python
print("Starting GridSearchCV fit...")
grid_search.fit(x_train_scaled, y_train)
print("GridSearchCV fit complete.")
```

**Reasoning**:
Retrieve and display the best parameters found by `GridSearchCV` and its corresponding best score. This will show the optimal hyperparameters for the Random Forest model.



```python
print("Best parameters found: ", grid_search.best_params_)
print("Best cross-validation score: ", grid_search.best_score_)
```

## Evaluate Optimized Random Forest Model

Evaluate the optimized Random Forest model using the test data to generate a classification report and visualize its confusion matrix. This will demonstrate the performance improvements after hyperparameter tuning.


**Reasoning**:
Retrieve the best estimator from the `grid_search` object, make predictions on the `x_test_scaled` data, print the classification report, and generate a confusion matrix.



```python
best_rf_model = grid_search.best_estimator_

print("--- Optimized Random Forest Model Evaluation ---")
y_pred_optimized_rf = best_rf_model.predict(x_test_scaled)
print("Classification Report:")
print(classification_report(y_test, y_pred_optimized_rf))

plt.figure(figsize=(12, 8))
sns.heatmap(confusion_matrix(y_test, y_pred_optimized_rf), annot=True, fmt='d', cmap='Purples',
            xticklabels=best_rf_model.classes_, yticklabels=best_rf_model.classes_)
plt.title('Confusion Matrix - Optimized Random Forest')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

## Retrieve Best Model and Scaler

Retrieve the best performing model from the GridSearchCV output (the optimized Random Forest model) and ensure the StandardScaler object used for training is available for new predictions.


**Reasoning**:
I need to retrieve the best model from the `grid_search` object and make sure the `scaler` is available. The `best_estimator_` attribute of `grid_search` holds the best model.



```python
optimized_rf_model = grid_search.best_estimator_
print("Optimized Random Forest Model retrieved successfully.")

# Confirming the scaler is available
if 'scaler' in locals():
    print("StandardScaler object is available for use.")
else:
    print("StandardScaler object is NOT available. Please re-run the scaling step if needed.")
```


Define a function `predict_crop` that takes the optimized Random Forest model, the scaler, and input data. Inside the function, it will scale the input data using the `scaler` and then use the `model` to predict the crop label. Finally, it will return the predicted label.



```python
def predict_crop(model, scaler, input_data_df):
    # Scale the input data using the fitted scaler
    scaled_input = scaler.transform(input_data_df)

    # Make prediction using the optimized model
    predicted_label = model.predict(scaled_input)

    return predicted_label[0]

print("The 'predict_crop' function has been defined and is ready for use.")
```

## Save Optimized Model and Scaler

Save the `optimized_rf_model` and the `StandardScaler` to disk using `pickle`.


**Reasoning**:
I need to save the `optimized_rf_model` and the `scaler` to disk. This involves importing the `pickle` library, defining filenames for both objects, and then using `pickle.dump()` to save them to separate files.



```python
import pickle

# Define filenames for the optimized model and scaler
optimized_model_filename = 'optimized_rf_model.pkl'
scaler_filename = 'scaler.pkl'

# Save the optimized Random Forest model
with open(optimized_model_filename, 'wb') as file:
    pickle.dump(optimized_rf_model, file)
print(f'Optimized Random Forest Model saved as {optimized_model_filename}')

# Save the StandardScaler object
with open(scaler_filename, 'wb') as file:
    pickle.dump(scaler, file)
print(f'StandardScaler object saved as {scaler_filename}')
```
